```python
import pandas as pd
import numpy as np
import time
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
from fcmeans import FCM
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('All_pecanoy_layer.txt', sep="\t")
```

```python
#df.rows = df.iloc[:,0]
df = df.set_index('gid')
```

```python
#dftem = df.iloc[0:10000,:]

start_time = time.time()

cosdf = cosine_similarity(df)

print("--- %s seconds ---" % (time.time() - start_time))

cosdf
```

```python
pd.DataFrame(cosdf, columns = df.index , index = df.index).to_csv("CosMatrix_All_layer.txt", sep='\t')
```

```python
cosdf.shape
```

```python
n_samples = 3000

X = np.concatenate((
    np.random.normal((-2, -2), size=(n_samples, 2)),
    np.random.normal((2, 2), size=(n_samples, 2))
))
```

```python
X = df.to_numpy()
X
```

```python
fcm = FCM(n_clusters=100)

fcm.fit(df.to_numpy())
```

```python
# outputs
fcm_centers = fcm.centers
fcm_labels = fcm.predict(df.to_numpy())

```

```python
# plot result
f, axes = plt.subplots(1, 2, figsize=(11,5))
axes[0].scatter(X[:,0], X[:,1], alpha=.1)
axes[1].scatter(X[:,0], X[:,1], c=fcm_labels, alpha=.1)
axes[1].scatter(fcm_centers[:,0], fcm_centers[:,1], marker="+", s=500, c='w')
#plt.savefig('images/basic-clustering-output.jpg')
plt.show()
```

```python
# make df with labes and gene ids
out = pd.DataFrame({'GeneID':df.index, 'Module':fcm_labels}) 
out.to_csv("CmeanGroupsPy.txt", sep='\t')
```
