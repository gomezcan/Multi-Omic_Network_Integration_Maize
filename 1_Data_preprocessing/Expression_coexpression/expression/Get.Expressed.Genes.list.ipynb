## Get genes expressed by network

```python
# Importing the os library
import os
import pandas as pd
```

```python
# read files by network

items = os.listdir('cpm_tables_raw')

my_df  = pd.DataFrame(columns=['netid', 'Geneid'])


# gid
for f in items[0:2]: 
    tem = pd.read_csv("cpm_tables_raw/"+f, delimiter="\t")
    #
    values = tem.iloc[:, 1:tem.shape[1]]
    #
    mask = (values > 3)*1

    mask = mask.sum(axis=1)
    #print(mask)    
    filter_value = tem.shape[1]*0.1 # at least 10% of the samples
    print(filter_value)    
    print ("Genes expressed")
    print((mask > filter_value).value_counts()[1])
    
    tem = tem[mask > filter_value]

    dftem = pd.DataFrame({'netid':f, "Geneid":tem.gid.values})    
    print(dftem)
    my_df = my_df.append(dftem)
    
    

my_df
```
